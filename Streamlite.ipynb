```python
%%writefile app.py

import streamlit as st

st.title("Student Marks Calculator")
st.write("Enter your marks to calculate your result.")

name = st.text_input("Enter Student Name")

st.subheader("Enter Subject Marks")

maths = st.number_input("Mathematics", 0, 100)
physics = st.number_input("Physics", 0, 100)
chemistry = st.number_input("Chemistry", 0, 100)
computer = st.number_input("Computer Science", 0, 100)
english = st.number_input("English", 0, 100)

if st.button("Calculate Result"):

    if name.strip() == "":
        st.error("Please enter student name.")

    else:
        marks = [maths, physics, chemistry, computer, english]

        total = sum(marks)
        percentage = total / 5

        st.subheader("Student Result")
        st.write("Student Name:", name)
        st.write("Total Marks:", total, "/ 500")
        st.write("Percentage:", percentage, "%")

        if min(marks) < 35:
            st.error("Result: FAIL")
            st.write("Grade: F")

        else:
            st.success("Result: PASS")

            if percentage >= 90:
                grade = "A+"
            elif percentage >= 80:
                grade = "A"
            elif percentage >= 70:
                grade = "B"
            elif percentage >= 60:
                grade = "C"
            elif percentage >= 50:
                grade = "D"
            else:
                grade = "E"

            st.write("Grade:", grade)
```
